You are given a string you need to reverse it

```python
# taking a string as an input.
def reverse_string(string_input):
    x = len(string_input)
    rev_string = ""
    for i in range(x,0,-1):
        rev_string += string_input[i-1]
    return rev_string

string_input = input("Enter the string")
y = reverse_string(string_input)
print(y)
```
